# **MINI CHALLENGE(SETS + LISTS + FUNCTION) : STUDENT ANALYZER SYSTEM**

```python
marks = [45, 67, 89, 45, 90, 67, 100, 89]

##REMOVES DUPLICATES
unique_marks = set(marks)

#PRINT UNIQUE MARKS
print(unique_marks)

##PRINT NUMBER OF UNIQUE MARKS
print(len(unique_marks))

## CREATE A FUNCTION (Fixed)
def analyze_marks(marks_list):
  """we use set to remove the duplicate values from the marks"""
  unique_marks2 = set(marks_list)
  return max(unique_marks2), min(unique_marks2)

# This will now correctly analyze your original 'marks' list
print("Max and Min from original list:", analyze_marks(marks))


## CHECK TOPPER
if 100 in marks:
    print("TOPPER")
else:
    print("NOT TOPPER")
print(analyze_marks.__doc__)



```

# **ULTIMATE QUESTION: STUDENT RESULT SYSTEM**

```python
try:
    name = input("Enter student name: ")
    physics = int(input("Enter physics marks: "))
    chemistry = int(input("Enter chemistry marks: "))
    maths = int(input("Enter maths marks: "))

    # Validate individual subject marks
    if not (0 <= physics <= 100 and 0 <= chemistry <= 100 and 0 <= maths <= 100):
        raise ValueError("Invalid marks: Marks must be between 0 and 100 for each subject.")

    # total
    total_marks = physics + chemistry + maths
    # percentage
    percentage = (total_marks / 3)
    # grade logic
    if percentage >= 90:
        print("Grade: A")
    elif percentage >= 80 and percentage < 90:
        print("Grade: B")
    elif percentage >= 70 and percentage < 80:
        print("Grade: C")
    elif percentage >= 60 and percentage < 70:
        print("Grade: D")
    else:
        print("Grade: F")

except ValueError as e:
    print(e)

else:
    print(f"Student name: {name}")
    print(f"Physics marks: {physics}")
    print(f"Chemistry marks: {chemistry}")
    print(f"Maths marks: {maths}")

finally:
    print("PROGRAM FINISHED")
```
