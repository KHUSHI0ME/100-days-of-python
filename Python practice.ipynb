```python
## FILTER FUNCTION
l = [10,15,20,25,30]
def filter_function(item):
  if item % 2 == 0:
    return True
  return False

filter_list = filter(filter_function,l)
print(list(filter_list))
```

```python
## Reduce Function
from functools import reduce
l = [1,2,3,4,5]
product = reduce(lambda x,y: x*y,l)
print(product)
```

```python
def greet(fx):
  def mfx():
   print("Good Morning, Hope you're doing well in your journey")
   fx()
   print("Hope you have a good day")
   # Removed 'return mfx' to allow implicit None return
  return mfx

@greet
def hello():
  print("hello world")

hello()
```

```python
class student:
  def __init__(self,marks):
    self._marks = marks

  def get_marks(self):
    return self._marks

  def set_marks(self,marks):
    if marks>0 and marks<=100:
      self._marks = marks

    else:
      print("Invalid marks")

student = student(90)

print(student.get_marks())

student.set_marks(95)

print(student.get_marks())
```

```python
class Employee:
  def __init__(self,name,id,post):
    self.name = name
    self.id = id
    self.post = post

  def set_salary(self,salary):
    self.salary = salary

obj1 = Employee("Khushi",201,"HR")
print(f"Name: {obj1.name}")
print(f"ID: {obj1.id}")
print(f"Post: {obj1.post}")



obj1.set_salary(50000)
print(f"Salary: {obj1.salary}")
```

```python
def welcome_decorator(fx):

    def wrapper():
        print("GOOD MORNING SIR, HOW CAN I HELP YOU TODAY")
        fx()
        print("HAVE A NICE DAY")

    return wrapper



class FinanceManager:

    def __init__(self, username, balance):
        self.username = username
        self.balance = balance
        self.expenses = []


    def add_expense(self, amount):
        self.expenses.append(amount)


    def view_expenses(self):
        print(self.expenses)


    def total_expense(self):
        return sum(self.expenses)


    def get_balance(self):
        return self.balance


    def set_balance(self, balance):

        if balance >= 0:
            self.balance = balance
        else:
            raise ValueError("Invalid Balance")


    def save_report(self):

        file = open("expense_report.txt","w")

        file.write(str(self.expenses))

        file.close()



@welcome_decorator
def Expenses():
    print("Here's your expenses sir")


Expenses()


manager = FinanceManager("Alice",1000)

manager.add_expense(500)
manager.add_expense(200)

print(manager.total_expense())
print(manager.get_balance())
```
