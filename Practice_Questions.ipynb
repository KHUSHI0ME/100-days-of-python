```python
## COUNT SPECIAL CHARACTERS

password = "Hello123@"

# For check every character
special_char_count = 0
for char in password:
  char.isalnum()

if not char.isalnum():
  special_char_count += 1

print(special_char_count)
```

```python
## Password Strength Checker

password = "Khushi123"
has_digit = False
has_upper = False

for char in password:
  if char.isdigit():
    has_digit = True
  elif char.isupper():
    has_upper = True

if has_digit and has_upper:
  print("Strong Password")
else:
  print("Weak Password")
```

```python
## WORDS FREQUENCY CHAMPION
sentence = "hello hello world"
words = sentence.split()
frequency = {}

for words in words:
  if words in frequency:
    frequency[words] += 1
  else:
    frequency[words] = 1

champion = max(frequency, key=frequency.get)

print("Word Frequency:", frequency)
print("Frequency Champion:", champion)


```

```python
list_num = [1, 2, 3, 2, 4, 1]

seen = []
duplicates = []

for num in list_num:
    if num not in seen:
        seen.append(num)
    else:
        if num not in duplicates:
            duplicates.append(num)

print(duplicates)
```

```python
## NUMBER SIGNATURE
num = int(input("Enter a number: "))

sum_digits = 0
product_digits = 1

while num > 0:
    digit = num % 10
    sum_digits += digit
    product_digits *= digit
    num = num // 10

print("Sum of digits:", sum_digits)
print("Product of digits:", product_digits)
```

```python
expenses = []

while True:
    print("\n--- SMART EXPENSE TRACKER ---")
    print("1. Add Expense")
    print("2. Show Total Expense")
    print("3. Show Highest Expense")
    print("4. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:
        amount = int(input("Enter expense amount: "))
        expenses.append(amount)
        print("Expense added!")

    elif choice == 2:
        print("Total Expense:", sum(expenses))

    elif choice == 3:
        if expenses:
            print("Highest Expense:", max(expenses))
        else:
            print("No expenses added yet.")

    elif choice == 4:
        print("Exiting... Thank you!")
        break

    else:
        print("Invalid choice!")
```
